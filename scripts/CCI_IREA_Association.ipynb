```python
Background_size = len(set(cci_df['ligand_complex']).union(set(cci_df['receptor_complex'])).union(set(MTC_456_irea_df["Cytokine"])))
# Fisher's Exact Test
a = len(mtc_123_significant_cci & MTC_123_significant_irea)  # Overlap
b = len(mtc_123_significant_cci - MTC_123_significant_irea)  # Only in CCI
c = len(MTC_123_significant_irea - mtc_123_significant_cci)  # Only in IREA
d = Background_size - (a + b + c)  # Background size (adjust if needed)

fet_pval = stats.fisher_exact([[a, b], [c, d]])[1]
fet_pval
```

```python
# Fisher's Exact Test
a = len(mtc_456_significant_cci & MTC_456_significant_irea)  # Overlap
b = len(mtc_456_significant_cci - MTC_456_significant_irea)  # Only in CCI
c = len(MTC_456_significant_irea - mtc_456_significant_cci)  # Only in IREA
d = Background_size - (a + b + c)  # Background size (adjust if needed)

fet_pval = stats.fisher_exact([[a, b], [c, d]])[1]
fet_pval
```
